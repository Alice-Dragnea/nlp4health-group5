```python
import os
import io
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
#https://www.geeksforgeeks.org/python/python-os-pardir-method-with-example/

path = os.getcwd() #path should be Evaluatio folder
parent = os.path.join(path, os.pardir) #parent directory
memo_dataset_path = os.path.join(parent, "Data/Memo_Dataset.csv")
#memo_dataset_path = os.path.abspath('Memo_Dataset.csv') 
print(memo_dataset_path)
```

```python
#loading and splitting the dataset 80/10/10
df_memo = pd.read_csv(memo_dataset_path, skiprows =0)
print(df_memo['Trigger'].value_counts())
```

We downsample before splitting our dataset to make sure there's an even representation of triggering and non triggering classes!

```python
df_majority = df_memo[df_memo['Trigger'] == 1].sample(frac=1, random_state=67)  
df_minority = df_memo[df_memo['Trigger'] == 0].sample(frac=1, random_state=67)  

# randomly downsampled from Trigger = 1 to match size of minority class Trigger = 0
df_majority_downsampled = df_majority.sample(n=len(df_minority), random_state=67)
df_memo_balanced = pd.concat([df_majority_downsampled, df_minority]).sample(frac=1, random_state=67) 
print(df_memo_balanced['Trigger'].value_counts(normalize=True))  # should show ~0.5/0.5
```

```python
#X_arb_eng = df_memo[['Question','Question_eng' ]]
X_arb = df_memo_balanced[['Question']]
X_eng = df_memo_balanced[['Question_eng']]
y = df_memo_balanced[['Trigger']].astype(int)

print(X_arb.shape)
print(X_eng.shape)
print(y.shape)
```

```python
#stratified to preserve  50/50 trigger representation from df_memo_balanced
X_arb_train, X_arb_validate_test, y_train, y_validate_test = train_test_split(X_arb, y, test_size=0.2, random_state = 67, stratify = y) #fixed seed so we get the same shuffling/samples when rerunning
X_eng_train, X_eng_validate_test, y_train, y_validate_test = train_test_split(X_eng y, test_size=0.2, random_state = 67, stratify = y)
print(X_arb_validate_test.shape)
print(y_validate_test.shape)
X_arb_validate, X_arb_test, y_validate, y_test = train_test_split(X_arb_validate_test, y_validate_test, test_size=0.5, random_state = 67, stratify = y_validate_test)
X_eng_validate, X_eng_test, y_validate, y_test = train_test_split(X_eng_validate_test, y_validate_test, test_size=0.5, random_state = 67, stratify = y_validate_test)
```
